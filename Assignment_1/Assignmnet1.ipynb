Part A: Dataset Understanding 

1.Brief Description of the Dataset

The Iris dataset contains information about iris flowers. It includes measurements of the sepal and petal of each flower. Using these measurements, the flowers are classified into three different species. This dataset is commonly used for learning data analysis and machine learning because it is simple and easy to understand.

2.Dataset Details
Dataset Source: Iris Dataset (UCI Machine Learning Repository)
Number of Samples: 150
Number of Features: 4
Target Variable: Species
Type of Problem: Classification

Features in the Dataset:
Sepal Length
Sepal Width
Petal Length
Petal Width

The target variable is Species, which contains three classes:
Setosa
Versicolor
Virginica
Since the dataset is used to predict the species of a flower based on its measurements, it is a classification problem.

Part C: Data Visualisation 

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

1.Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(df["sepal_length"], bins=10, edgecolor="black")
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

2.There are fewer flowers with very small or very large sepal lengths.

3.The data is concentrated around the middle values.

2.Box Plot

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["sepal_length"])
plt.title("Box Plot of Sepal Length")
plt.show()
```

1.Most values lie within the middle range of the data.

2.A few outliers can be seen outside the box.

3.Scatter Plot

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species"
)

plt.title("Sepal Length vs Petal Length")
plt.show()
```

4.Pair Plot

```python
sns.pairplot(
    df,
    vars=["sepal_length", "sepal_width", "petal_length", "petal_width"],
    hue="species"
)

plt.show()
```

2.Setosa forms a distinct cluster.

3.Versicolor and virginica overlap slightly.

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df.drop("species", axis=1).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

Part E: Data Preprocessing

Handle Missing Values

```python
print(df.isnull().sum())

df.fillna(df.mean(numeric_only=True), inplace=True)

print("Missing values after handling:")
print(df.isnull().sum())
```

Justification:

First, the dataset was checked for missing values. Since no missing values were found, no changes were required. If any had been present, they could have been replaced with the column mean.

Remove or Treat Outliers

```python
Q1 = df.quantile(0.25, numeric_only=True)
Q3 = df.quantile(0.75, numeric_only=True)
IQR = Q3 - Q1

df_no_outliers = df[
    ~((df.select_dtypes(include='number') < (Q1 - 1.5 * IQR)) |
      (df.select_dtypes(include='number') > (Q3 + 1.5 * IQR))).any(axis=1)
]

print("Original Shape:", df.shape)
print("After Removing Outliers:", df_no_outliers.shape)
```

Justification:

The IQR method was used to check for outliers. This helps identify values that are very different from the rest of the data. Only values outside the normal range are removed.

Encode Categorical Variables

```python
# Convert species into numerical values
df["species"] = df["species"].astype("category").cat.codes

print(df.head())
```

Justification:

The species column contains text values. It was converted into numerical values because machine learning algorithms work with numbers instead of text.

Standardize Numerical Features

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numerical_columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

df[numerical_columns] = scaler.fit_transform(df[numerical_columns])

print(df.head())
```

Justification:

The numerical columns were standardized so that all features are on a similar scale. This prevents one feature from having more influence than the others.

Part F: Sampling

Simple Random Sampling

```python
random_sample = df.sample(frac=0.30, random_state=42)

print(random_sample.head())
print("Sample Size:", len(random_sample))
```

Justification

A simple random sample was created so that every record had an equal chance of being selected This helps reduce selection bias

Stratified Sampling

```python
stratified_sample = (
    df.groupby("species", group_keys=False)
      .apply(lambda x: x.sample(frac=0.30, random_state=42))
)

print(stratified_sample.head())
print("Sample Size:", len(stratified_sample))
```

Justification:

Stratified sampling was used to ensure that each flower species was represented in the sample. This maintains the original class distribution

Descriptive Statistics of the Random Sample

```python
print("Simple Random Sample Statistics")
print(random_sample.describe())
```

Descriptive Statistics of the Stratified Sample

```python
print("Stratified Sample Statistics")
print(stratified_sample.describe())
```

Comparison of Both Samples

The mean values of both samples are very close to each other

The standard deviation is also similar, showing that both samples represent the original dataset well

The stratified sample includes records from every species, while the simple random sample may have slightly different numbers of each species

Overall stratified sampling provides a more balanced representation of the dataset
